## Part 6: Download Language Models

````python
# Pre-download models
# This downloads all compatible models for your hardware
# Takes 5-30 minutes depending on internet speed and models

from src.models.registry import get_compatible_models, MODELS
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
import gc

# Detect compatible models for your hardware
device = torch.device("cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu")

# Check if 8-bit quantization is configured (from Part 6 config)
# If not defined yet, default to True for better compatibility
try:
    use_8bit_for_download = USE_8BIT
except NameError:
    use_8bit_for_download = True
    print("ℹ️  USE_8BIT not set yet, defaulting to True for compatibility check\n")

compatible_models = get_compatible_models(use_8bit=use_8bit_for_download)

print(f"🔍 Detected device: {device}")
print(f"🔧 8-bit quantization: {'Enabled' if use_8bit_for_download else 'Disabled'}")
print(f"📦 Found {len(compatible_models)} compatible models")
print(f"\nModels to download:")
for key in compatible_models:
    vram_needed = MODELS[key].vram_8bit_gb if use_8bit_for_download else MODELS[key].vram_gb
    print(f"  • {MODELS[key].name} ({MODELS[key].params}) - {vram_needed:.1f}GB with {'8-bit' if use_8bit_for_download else 'no'} quantization")

print(f"\n⚠️  This will download {len(compatible_models)} models (can be several GB)")
print("   Press Ctrl+C to cancel, or wait to continue...\n")

import time
time.sleep(3)

# Helper function for aggressive memory cleanup
def cleanup_memory():
    """Aggressive memory cleanup for Apple Silicon."""
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()
    elif torch.backends.mps.is_available():
        torch.mps.empty_cache()
        gc.collect()  # Second pass for MPS

# Download each model
for i, model_key in enumerate(compatible_models, 1):
    model_config = MODELS[model_key]
    print(f"\n[{i}/{len(compatible_models)}] Downloading {model_config.name}...")
    print(f"    HuggingFace ID: {model_config.hf_id}")
    
    try:
        # Download tokenizer and model
        tokenizer = AutoTokenizer.from_pretrained(model_config.hf_id, trust_remote_code=True)
        
        # Use low_cpu_mem_usage to avoid loading into memory
        model = AutoModelForCausalLM.from_pretrained(
            model_config.hf_id,
            trust_remote_code=True,
            low_cpu_mem_usage=True,
            dtype=torch.float16 if device != "cpu" else torch.float32
        )
        
        # Immediately free memory
        del model
        del tokenizer
        
        # Aggressive cleanup after each download
        cleanup_memory()
        
        print(f"    ✅ Downloaded successfully")
        
    except Exception as e:
        print(f"    ❌ Failed: {e}")
        # Cleanup even on failure
        cleanup_memory()

# Final cleanup
cleanup_memory()

print(f"\n🎉 Model download complete!")
print("   You can now run Phase 3 without interruptions.")
````

````python
# Phase 2 (Batch): All Categories evaluation (safe to interrupt)
## Tests only: zero_shot vs few_shot (all curated examples)
## Saves per-run artifacts to data/experiments/ via ExperimentTracker
## Also saves a summary CSV for quick overview

from pathlib import Path
from datetime import datetime
import json

import numpy as np
import pandas as pd

from src.classification import TextClassifier
from src.classification.prompts import CATEGORIES, list_prompt_templates, get_prompt
from src.evaluation.metrics import calculate_metrics
from src.utils import ExperimentTracker

print("Starting Phase 2 (Batch): Template Comparison (ALL categories)\n")

# Discover categories from available test data
TEST_DIR = Path("data/processed/test")
if not TEST_DIR.exists():
    raise FileNotFoundError(f"Missing test directory: {TEST_DIR}")

categories = sorted([p.stem for p in TEST_DIR.glob("*.csv")])
if not categories:
    raise FileNotFoundError(f"No test CSVs found under: {TEST_DIR}")

# Templates to test (display_name -> template_key)
templates = list_prompt_templates()

print(f"Found {len(categories)} categories in {TEST_DIR}")
print("Categories:", ", ".join(categories))
print(f"Templates: {', '.join(templates.keys())}\n")

# Summary CSV path (for quick overview across all runs)
summary_dir = Path("results/phase2_batch_all_categories")
summary_dir.mkdir(parents=True, exist_ok=True)

ts = datetime.now().strftime("%Y%m%d_%H%M%S")
model_slug = (
    str(MODEL)
    .replace("/", "__")
    .replace("\\", "__")
    .replace(":", "_")
    .replace(" ", "_")
)

summary_csv_path = summary_dir / f"phase2_summary__{model_slug}__{ts}.csv"
print("Summary CSV:", summary_csv_path)
print("Per-run artifacts: data/experiments/experiment_<timestamp>__<category>__<exp_id>/")
print()

# ===========================
# PHASE 1: Check what work is needed (no model loading yet!)
# ===========================
print("=" * 80)
print("PHASE 1: Checking experiment cache...")
print("=" * 80)

experiments_to_run = []  # List of (category, category_index, template_name, template_key, test_path, df)
cached_results = []  # List of cached result dicts for summary

for ci, category in enumerate(categories, 1):
    test_path = TEST_DIR / f"{category}.csv"
    df = pd.read_csv(test_path)
    if MAX_SAMPLES:
        df = df.head(int(MAX_SAMPLES))
    
    # Create tracker for this category
    tracker = ExperimentTracker(category)
    
    for ti, (template_name, template_key) in enumerate(templates.items(), 1):
        # Check if experiment already exists
        check_result = tracker.check_duplicate(
            model=MODEL,
            prompt_template=template_key,
            max_samples=MAX_SAMPLES,
            use_8bit=USE_8BIT,
            verbose=False,
        )
        
        if check_result['cached']:
            # Already have results
            cached = check_result['cached']
            metrics = cached.get('metrics', {})
            samples_skipped = cached['dataset']['total_samples'] - cached['dataset']['samples_used']
            
            cached_results.append({
                "category": category,
                "template_name": template_name,
                "template_key": template_key,
                "metrics": metrics,
                "samples_processed": int(cached['dataset']['samples_used']),
                "samples_skipped": int(samples_skipped),
            })
            print(f"✓ [{ci}/{len(categories)}] {category} | {template_name}: CACHED")
        else:
            # Need to run this experiment
            experiments_to_run.append({
                "category": category,
                "category_index": ci,
                "template_name": template_name,
                "template_key": template_key,
                "test_path": test_path,
                "df": df,
                "tracker": tracker,
            })
            print(f"○ [{ci}/{len(categories)}] {category} | {template_name}: NEEDS RUN")

print()
print(f"Cache scan complete: {len(cached_results)} cached, {len(experiments_to_run)} to run")
print()

# If everything is cached, skip model loading entirely!
if not experiments_to_run:
    print("🎉 All experiments already cached! No model loading needed.")
    print()
    print("All results are available in: data/experiments/")

else:
    # ===========================
    # PHASE 2: Load model and run experiments
    # ===========================
    print("=" * 80)
    print("PHASE 2: Loading model and running experiments...")
    print("=" * 80)
    print()
    
    # Load model ONCE and reuse across all categories for speed
    classifier = TextClassifier(
        model_name=MODEL,
        use_8bit=USE_8BIT,
        batch_size=2,
    )

    # Detect format from classifier
    use_json_format = classifier.use_json

    # Resolve MODEL to HuggingFace ID for consistent logging
    from src.models import get_model_config
    try:
        model_config = get_model_config(MODEL)
        MODEL_HF_ID = model_config.hf_id
    except:
        MODEL_HF_ID = MODEL  # Fallback if not in registry

    print(f"Output format: {'JSON' if use_json_format else 'Text'} (auto-detected from model)")
    print(f"Model HF ID: {MODEL_HF_ID}")
    print()

    # Summary rows: start with cached results
    summary_rows = []
    for cached in cached_results:
        summary_rows.append({
            "category": cached["category"],
            "template_name": cached["template_name"],
            "template_key": cached["template_key"],
            "output_format": "json" if use_json_format else "text",
            "accuracy": float(cached["metrics"].get("accuracy")) if cached["metrics"].get("accuracy") is not None else None,
            "qwk": float(cached["metrics"].get("qwk")) if cached["metrics"].get("qwk") is not None else None,
            "mse": float(cached["metrics"].get("mse")) if cached["metrics"].get("mse") is not None else None,
            "mae": float(cached["metrics"].get("mae")) if cached["metrics"].get("mae") is not None else None,
            "samples_processed": cached["samples_processed"],
            "samples_skipped": cached["samples_skipped"],
            "error": "",
        })

    def save_summary():
        pd.DataFrame(summary_rows).to_csv(summary_csv_path, index=False, encoding="utf-8")

    try:
        for exp_idx, exp in enumerate(experiments_to_run, 1):
            category = exp["category"]
            ci = exp["category_index"]
            template_name = exp["template_name"]
            template_key = exp["template_key"]
            df = exp["df"]
            tracker = exp["tracker"]
            
            print("=" * 80)
            print(f"[{exp_idx}/{len(experiments_to_run)}] Running: {category} | {template_name}")
            print("=" * 80)

            if "free_text_answer" not in df.columns:
                raise KeyError(f"Missing 'free_text_answer' column in {category}")
            if category not in df.columns:
                raise KeyError(f"Missing label column '{category}' in {category}")

            texts = df["free_text_answer"].astype(str).tolist()
            true_labels = df[category].tolist()

            try:
                # Get duplicate check result again (for metadata)
                check_result = tracker.check_duplicate(
                    model=MODEL,
                    prompt_template=template_key,
                    max_samples=MAX_SAMPLES,
                    use_8bit=USE_8BIT,
                    verbose=True,
                )
                
                # Run the classification
                pred_results = classifier.classify_batch(
                    category=category,
                    texts=texts,
                    prompt_template=template_key,
                    return_raw=False,
                    show_progress=True,
                )

                y_pred = [r["score"] for r in pred_results if r.get("score") is not None]
                y_true = [true_labels[i] for i, r in enumerate(pred_results) if r.get("score") is not None]

                y_true_arr = np.array(y_true)
                y_pred_arr = np.array(y_pred)

                metrics = calculate_metrics(y_true_arr, y_pred_arr)
                samples_skipped = len(texts) - len(y_pred)

                # Build predictions list for tracker (expected format)
                predictions_list = []
                for idx, r in enumerate(pred_results):
                    predictions_list.append({
                        "sample_index": idx,
                        "text": texts[idx],
                        "y_true": true_labels[idx],
                        "y_pred": r.get("score"),
                        "parse_ok": r.get("score") is not None,
                        "raw_output": r.get("raw_output", ""),
                    })

                # Log to ExperimentTracker (creates per-run folder)
                tracker.log_experiment(
                    model=check_result['model_hf_id'],
                    prompt_template=template_key,
                    prompt_text=check_result['prompt_text'],
                    metrics=metrics,
                    dataset_path=check_result['dataset_path'],
                    total_samples=len(df),
                    samples_used=len(y_pred),
                    generation_params=classifier.get_generation_params(),
                    output_format=check_result['output_format'],
                    max_samples=MAX_SAMPLES,
                    quantization=check_result['quantization'],
                    version_name=f"phase2_batch_{template_name}",
                    notes=f"Phase 2 batch run: {template_name}",
                    predictions=predictions_list,
                )

                # Add to summary
                summary_rows.append({
                    "category": category,
                    "template_name": template_name,
                    "template_key": template_key,
                    "output_format": "json" if use_json_format else "text",
                    "accuracy": float(metrics.get("accuracy")) if metrics.get("accuracy") is not None else None,
                    "qwk": float(metrics.get("qwk")) if metrics.get("qwk") is not None else None,
                    "mse": float(metrics.get("mse")) if metrics.get("mse") is not None else None,
                    "mae": float(metrics.get("mae")) if metrics.get("mae") is not None else None,
                    "samples_processed": int(len(y_pred)),
                    "samples_skipped": int(samples_skipped),
                    "error": "",
                })

                print(
                    f"✓ OK Accuracy={float(metrics['accuracy']):.1%}, QWK={float(metrics['qwk']):.3f}"
                    + (f" ({samples_skipped} failed parses)" if samples_skipped else "")
                )

            except Exception as e:
                err = str(e)
                summary_rows.append({
                    "category": category,
                    "template_name": template_name,
                    "template_key": template_key,
                    "output_format": "json" if use_json_format else "text",
                    "accuracy": None,
                    "qwk": None,
                    "mse": None,
                    "mae": None,
                    "samples_processed": 0,
                    "samples_skipped": len(texts),
                    "error": err,
                })
                print(f"✗ FAILED: {err}")

            # Reduce memory growth across many runs
            classifier.clear_cache()

            # Save summary after each experiment
            save_summary()

        print()
        print("=" * 80)
        print("Phase 2 (batch all categories) completed!")
        print("=" * 80)
        print("Saved summary:", summary_csv_path)
        print("Per-run artifacts in: data/experiments/")

    except KeyboardInterrupt:
        print("\nInterrupted by user. Saving partial summary...")
        try:
            save_summary()
            print("Saved:", summary_csv_path)
        finally:
            raise
    finally:
        # Ensure GPU memory is freed
        try:
            classifier.unload()
        except Exception:
            pass
````

## Part 10: RAG Setup (Retrieval-Augmented Generation)

RAG enhances few-shot classification by **dynamically retrieving similar examples** from your training data for each input text, instead of using the same static examples for all inputs.

**What RAG does:**
- Embeds your training data into a vector store
- For each classification, finds the most similar examples
- Uses those relevant examples in the prompt

**Benefits over static few-shot:**
| Static Few-Shot | RAG Few-Shot |
|-----------------|--------------|
| Same examples for ALL inputs | Different examples per input |
| May not match the input context | Semantically similar to input |
| Limited to ~5 examples in prompt | Searches 1000s of examples |

````python
# Verify RAG dependencies are installed
print("Checking RAG dependencies...\n")

rag_packages = {
    "sentence_transformers": "Sentence-Transformers (embeddings)",
    "faiss": "FAISS (vector search)",
}

rag_ok = True
for pkg, description in rag_packages.items():
    try:
        __import__(pkg)
        print(f"✅ {description}")
    except ImportError:
        print(f"❌ {description} - NOT INSTALLED")
        rag_ok = False

if not rag_ok:
    print("\n⚠️  Missing RAG packages. Installing now...")
    import subprocess, sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "sentence-transformers", "faiss-cpu", "-q"])
    print("✅ RAG packages installed!")
else:
    print("\n🎉 All RAG dependencies are ready!")
````
